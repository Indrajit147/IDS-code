# M9 on Kaggle: full-pool E4, with the K sweep, the trigger passes and the adaptive gate

**Seed:** set `SEED` in the first code cell. Run 42 first, then repeat with 43 in a second session, as for M7. The partition stays the same, only the model seeds change.

**Attach as input:** the same private Kaggle dataset as `m3_kaggle.ipynb` (the five files of the M1 `processed/` folder).

**Settings:** Internet *On*. Accelerator: *None*.
**Run:** *Save Version → Save & Run All (Commit)* so it keeps running with the browser closed. An interactive session can stop when it is left alone.
Time is an estimate, not measured on this pool: eight passes at about 1 hour each, so allow 8 to 10 hours per seed. The session limit is 12 hours, which is why each seed gets its own run.

Each pass runs in its own process, and the zip is rewritten after each one, so a failure does not lose what finished. Every pass prints an exit code: 137 means the process was killed (usually out of memory). `--report` then writes the report from whatever finished, and says which passes are missing. `e4` and `e4_added` stop with an error if the K sweep isn't complete.

**Output:** `/kaggle/working/m9_full_seed42_outputs.zip` (the name follows `SEED`) contains `reports/m9/full/` and the run folders. Download it and bring it back, also after a failure.


```python
SEED = 42
TAG = "" if SEED == 42 else f"seed{SEED}"   # seed 42 is the main report, as in M7
TAG_ARG = f"--tag {TAG}" if TAG else ""
import os, subprocess, sys
REPO = "/kaggle/working/IDS-code"
if not os.path.isdir(REPO):
    subprocess.run(["git", "clone", "--depth", "1", "https://github.com/Indrajit147/IDS-code.git", REPO], check=True)
else:
    subprocess.run(["git", "-C", REPO, "pull", "--ff-only"], check=True)
# --no-deps: keep Kaggle's preinstalled stack intact; versions are recorded per run.
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-e", REPO, "--no-deps"], check=True)
os.chdir(REPO)
os.environ["IDS_REPO_ROOT"] = REPO
print(subprocess.run(["git", "log", "-1", "--oneline"], capture_output=True, text=True).stdout)
```

```python
import glob, os, shutil
REP = os.path.join("reports/m9/full", TAG)
def pack():
    stage = "/kaggle/working/m9_stage"
    shutil.rmtree(stage, ignore_errors=True)
    if os.path.isdir(REP):
        shutil.copytree(REP, f"{stage}/{REP}", ignore=shutil.ignore_patterns("checkpoints"))
    for r in sorted(glob.glob("/kaggle/working/results/runs/*_m9_e4_full")):
        shutil.copytree(r, f"{stage}/results/runs/{os.path.basename(r)}")
    shutil.make_archive(f"/kaggle/working/m9_full_seed{SEED}_outputs", "zip", stage)
    shutil.rmtree(stage)
    print("zip written:", sorted(os.listdir(REP)) if os.path.isdir(REP) else [])
```

```python
# All passes, each in its own process. 4 threads, because results reproduce exactly only at the same thread count
for name in ["fixed", "fixed_k10", "fixed_k20", "fixed_k30", "e4_schedule", "e4_label_free", "e4", "e4_added"]:
    print("=====", name, flush=True)
    !python -X faulthandler scripts/m9_e4.py --env kaggle --pool full --set resources.n_jobs=4 seed={SEED} {TAG_ARG} --only {name} --resume; echo "exit code: $?"
    pack()
```

```python
# Report from whatever finished
!python -X faulthandler scripts/m9_e4.py --env kaggle --pool full --set resources.n_jobs=4 seed={SEED} {TAG_ARG} --report; echo "exit code: $?"
pack()
if os.path.exists(f"{REP}/E4.md"):
    print(open(f"{REP}/E4.md", encoding="utf-8").read()[:6000])
```
