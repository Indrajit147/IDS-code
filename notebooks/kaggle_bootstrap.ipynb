# Kaggle bootstrap — IDS-code

**One-time Kaggle setup**
1. *Settings → Internet: On* (needed to clone the repo; requires a phone-verified account).
2. *Add Input*: attach your private dataset containing `ciciomt2024_full_8p7m.csv` and the Zenodo
   `extracted/` files (`bluetooth_clean.parquet`, `wifi_mqtt_clean.parquet`, ...). Files are found by
   name anywhere under `/kaggle/input`, so the dataset slug does not matter.
3. *Accelerator*: None for M1–M10; GPU T4/P100 for M11–M12 (LLM).

Every later Kaggle notebook starts with the two setup cells below.

```python
import os, subprocess, sys
REPO = "/kaggle/working/IDS-code"
if not os.path.isdir(REPO):
    subprocess.run(["git", "clone", "--depth", "1", "https://github.com/Indrajit147/IDS-code.git", REPO], check=True)
else:
    subprocess.run(["git", "-C", REPO, "pull", "--ff-only"], check=True)
# --no-deps: keep Kaggle's preinstalled stack intact; versions are recorded per run.
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-e", REPO, "--no-deps"], check=True)
os.chdir(REPO)
os.environ["IDS_REPO_ROOT"] = REPO
print(subprocess.run(["git", "log", "-1", "--oneline"], capture_output=True, text=True).stdout)
```
